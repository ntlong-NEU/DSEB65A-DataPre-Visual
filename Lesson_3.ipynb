```python
def group_method(planets):
    group_df={}
    group_df['Method'] = planets['method'].unique()
    group_df['Total'] = []
    for m in group_df['Method']:
        group_df['Total'].append(planets[planets['method']==m]['number'].sum())
    return pd.DataFrame(group_df)
    
```

```python
group_method(planets)
```

```python
url = 'https://raw.githubusercontent.com/jakevdp/data-CDCbirths/master/births.csv'
births = pd.read_csv(url)
births.head()
```

```python
df = pd.pivot_table(data=births,
                    index='year',
                    values='births',
                    columns='gender',
                    aggfunc='sum')
df.head()
```

```python
import seaborn as sns
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
# Create age groups
age_groups = pd.cut(titanic['age'],
                    bins=[0, 12, 18, 35, 60, 80], 
                    labels=['Child', 'Teen', 'Adult', 'Middle-aged', 'Senior'])
# Add age group column to the DataFrame
titanic['age_group'] = age_groups
titanic['age_group'].head()
```

```python
titanic['age_group'].value_counts()
```

```python
titanic['age_group'].isnull().sum()
```

```python
df_ageg_sex = pd.pivot_table(data=titanic, 
               index='age_group',
               columns='sex',
               values='survived',
               aggfunc='sum')
df_ageg_sex
```

```python
df_ageg_sex.plot.barh(title = 'Survived by age group and sex');
```
